```python
import pandas as pd

# 1. Загружаем сырые данные
df = pd.read_csv('../data/raw/Norway-NA-15_47_9-F-9 A depth.csv')

# 2. Задаем точные имена колонок из нашего датасета
columns_to_keep = [
    'Corrected Surface Weight on Bit kkgf', # WOB (Нагрузка на долото)
    'Average Rotary Speed rpm',             # RPM (Обороты ротора)
    'Total SPM 1/min',                      # Flow Rate (Расход жидкости в ходах насоса)
    'Average Surface Torque kN.m',          # Torque (Крутящий момент)
    'Rate of Penetration m/h'               # ROP (Скорость бурения - наш таргет)
]

# 3. Оставляем в датафрейме только эти 5 колонок
df = df[columns_to_keep]

# 4. Переименовываем их в короткие и удобные названия
df.columns = ['WOB', 'RPM', 'Flow Rate', 'Torque', 'ROP']

# 5. Очистка: удаляем пустые значения (NaN) и строки, где скорость нулевая или отрицательная
df_clean = df[df['ROP'] > 0].dropna()

# 6. Проверяем, сколько строк осталось после очистки
print(f"Размер очищенного датасета: {df_clean.shape}")

df_clean.to_csv('../data/processed/clean_drilling_data.csv', index=False)

# 7. Выводим первые 5 строк красивой таблицы
df_clean.head()
```
